### 记忆系统

```python
from langchain.memory import ConversationBufferMemory
from langchain import ConversationChain
```

```python
#创建一个记忆系统
mmo = ConversationBufferMemory()
#创建一个会话链
conversation = ConversationChain(llm=chat,memory=mmo)  #传递LLM的原因在于不同模型token消耗量不同
```

```python
conversation.predict(input="你好，我是一个学生")
```
